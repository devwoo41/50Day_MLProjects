#### CSV 파일의 구분자가 ';'으로 존재하는 이슈 해결

```python
import pandas as pd

red_wine_df = pd.read_csv('./datasets/winequality-red.csv', sep=';')
white_wine_df = pd.read_csv('./datasets/winequality-white.csv', sep=';')
print(red_wine_df.head())
print(white_wine_df.head())
```

#### 타겟 변수 분포 확인

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 제거 (기본 폰트 사용)
sns.set(style="whitegrid")

# 분석할 데이터셋 선택
df = red_wine_df.copy()

# -------------------------------------------------------
# 1. 타겟 변수(Quality) 분포 확인
# -------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.countplot(x='quality', data=df, palette='viridis')
plt.title('Target Distribution (Red-wine Quality)', fontsize=15) # 영어로 변경
plt.xlabel('Quality (Score)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

Red wine EDA 결과 : 5~6 점에 와인 데이터가 몰려있다

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 제거 (기본 폰트 사용)
sns.set(style="whitegrid")

# 분석할 데이터셋 선택
df = white_wine_df.copy()

# -------------------------------------------------------
# 1. 타겟 변수(Quality) 분포 확인
# -------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.countplot(x='quality', data=df, palette='viridis')
plt.title('Target Distribution (White-wine Quality)', fontsize=15) # 영어로 변경
plt.xlabel('Quality (Score)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

White wine EDA 결과 : RED wine에 비해 상대적으로 고점에 몰려있고, 평이하나 여전히 분포가 고르지 못하다

#### 상관관계 히트맵

```python
plt.figure(figsize=(12, 10))
df= red_wine_df.copy()
corr = df.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Matrix(Red-wine)', fontsize=15) # 영어로 변경
plt.show()
```

```python
plt.figure(figsize=(12, 10))
df= white_wine_df.copy()
corr = df.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Matrix(White-wine)', fontsize=15) # 영어로 변경
plt.show()
```
